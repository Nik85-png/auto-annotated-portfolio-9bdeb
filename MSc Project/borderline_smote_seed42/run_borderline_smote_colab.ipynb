# HF-RISK BorderlineSMOTE Experiment

This notebook copies the MSc Project into a fresh runtime folder, swaps in the BorderlineSMOTE experiment scripts, and runs the pipeline with seed 42.


```python
from pathlib import Path
import os
import shutil
import subprocess
import sys

project_root = None
for candidate in [Path.cwd(), *Path.cwd().parents]:
    if (candidate / 'MSc Project').exists():
        project_root = candidate / 'MSc Project'
        break
if project_root is None:
    raise FileNotFoundError('Could not find MSc Project folder.')

experiment_assets = project_root / 'borderline_smote_seed42'
runtime_root = project_root.parent / 'MSc Project_borderline_smote_seed42_runtime'
if runtime_root.exists():
    shutil.rmtree(runtime_root)
shutil.copytree(project_root, runtime_root)
shutil.copy2(experiment_assets / '02_preprocessing.py', runtime_root / '02_preprocessing.py')
shutil.copy2(experiment_assets / '03_modelling.py', runtime_root / '03_modelling.py')
print(f'Runtime copy created at: {runtime_root}')
print('Using seed 42 with BorderlineSMOTE.')
```

```python
subprocess.run([sys.executable, "-m", "pip", "install", "-r", str(project_root / "requirements.txt")], check=True)
```

```python
os.chdir(runtime_root)
print("Running preprocessing...")
subprocess.run([sys.executable, "02_preprocessing.py"], check=True)
print("Running modelling...")
subprocess.run([sys.executable, "03_modelling.py"], check=True)
```
